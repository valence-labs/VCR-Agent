## Entity retrieval for gene and compound

```python
%load_ext autoreload
%autoreload 2

from explain.eval.tools.bio.entity import CompoundEntity, FlairModel, GeneEntity
```

#### Specify a NER model

```python
ner_model = FlairModel()
```

#### Define the GeneEntity with text representation for a gene (gene name, synonym, short description)

```python
ge = GeneEntity(name="EGFR")
```

```python
# use a specified ner model for gene entity and identifier retrieval
ge.retrieve_identifiers(ner_model)
```

```python
# show all the identifiers for the query gene.
list(ge)
```

### Now use another text for the same gene

```python
text = "The EGFR gene provides instructions for making the epidermal growth factor receptor (EGFR) protein"

ge2 = GeneEntity(name=text)
```

```python
ge2.retrieve_identifiers()
```

```python
list(ge2)
```

### Compound entity retrieval
