#### Get labelled interactions between genes

Remember computing the id (ensembl id) and loading from the ensembl id is always better than trying to do it with the gene symbol

```python
from explain.utils.entity import GeneEntity
import pandas as pd

gene = await GeneEntity.from_name("JAK1")
gene2 = await GeneEntity.from_name("STAT3")
out = await gene.get_process_interaction(gene2)
print(out)
pd.DataFrame.from_records(out) # if you want to see the data as a dataframe
```
