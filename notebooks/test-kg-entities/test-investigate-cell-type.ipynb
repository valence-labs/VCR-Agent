### Investigation of cell context

```python
from google.cloud import bigquery

client = bigquery.Client(project="datalake-prod-ef49c0c9")

cell_t_list = [
    'iPSC-CORTICAL-ASTROCYTE','hMSC','HRCE','NHBE','HUVEC_BMP9','NHCE','A549','Other',
    'HUVEC_SF','SHSY5Y','AU565','RPEAd','BxPC3','HAEC','HEPARG','PTEC','iPSC-NGN2-NEURON',
    'LX-2','NHEK','SkMM','iCELL-CARDIOMYOCYTE','ASTROCYTE','PASMC','HACAT','AC16','NHCF',
    'VERO','CACO2','HUVEC_TNFa','ARPE19','FIBROBLAST','HEPG2','HUVEC','HUVEC_TNFalpha',
    'MACROPHAGE','MCF7','NHDF','NHLF','PBMC','RPE','RPTEC','U2OS','HMVECc'
]

query = """
SELECT
  cc.rec_id,
  ec.synonyms,
  ec.type,
  ec.fuzzy_representation,
  ctc.label AS cell_type,
  tc.label  AS tissue,
  sc.label  AS species,
  cc.culture_type,
  cc.disease,
  cc.vendor,
  cc.vendor_identifier,
  cc.vendor_lot,
  ec.tags,
  cc.created_at,
  cc.updated_at
FROM `cauldron__cauldron.cells`       AS cc
JOIN `cauldron__cauldron.entities`    AS ec ON ec.rec_id     = cc.rec_id
JOIN `cauldron__cauldron.cell_types`  AS ctc ON ctc.id       = cc.cell_type_id
JOIN `cauldron__cauldron.tissues`     AS tc  ON tc.id        = cc.tissue_id
JOIN `cauldron__cauldron.species`     AS sc  ON sc.id        = cc.species_id
WHERE EXISTS (
  SELECT 1
  FROM UNNEST(IFNULL(ec.synonyms, [])) AS s
  JOIN UNNEST(@terms) AS t
  ON STRPOS(LOWER(s), LOWER(t)) > 0
)
"""

job = client.query(
    query,
    job_config=bigquery.QueryJobConfig(
        query_parameters=[bigquery.ArrayQueryParameter("terms", "STRING", cell_t_list)]
    ),
)

df = job.result().to_dataframe()
df.head()
```
